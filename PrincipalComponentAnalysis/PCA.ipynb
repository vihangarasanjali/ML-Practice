```python
import matplotlib_inline
import numpy as np
import matplotlib.pyplot as plt

from Feature_Scaling.Normalization import scaled_data

matplotlib_inline
```

```python
data = np.array([[40,20],
                 [55,30],
                 [70,60],
                 [50,35],
                 [45,40],
                 [62,75],
                 [45,30],
                 [68,80],
                [80,70],
                 [75,90]])
```

```python
plt.scatter(data[:,0],data[:,1])
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data=scaler.fit_transform(data)
print(scaled_data)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=1)
pca.fit(scaled_data)
```

```python
pca.explained_variance_ratio_
```

```python
pca_scaled_data = pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
pca_scaled_data.shape
```

```python
pca_scaled_data = pca.inverse_transform(pca_scaled_data)
```

```python
plt.scatter(scaled_data[:,0],scaled_data[:,1])
plt.scatter(pca_scaled_data[:,0],pca_scaled_data[:,1])
```
