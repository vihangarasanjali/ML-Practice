```python
import pandas as pd
import matplotlib.pyplot as plt
from seaborn.matrix import dendrogram
```

```python
data = pd.DataFrame(data={
    'x': [0,1.1,1,2,2,4,5,5],
    'y': [0,1.5,4,2,3,1,0,4],
})
```

```python
data
```

```python
plt.scatter(data['x'], data['y'])
letters= ['A','B','C','D','E','F','G','H']
for index, value in enumerate(letters):
    plt.annotate(value, xy=(data['x'][index], data['y'][index]))
```

```python
import scipy.cluster.hierarchy as hc
dendrogram = hc.dendrogram(hc.linkage(data, method='ward'))
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
pred = model.fit_predict(data)
```

```python
pred
```

```python
data['Cluster'] = pred
data
```

```python
cluster0 = data[data['Cluster']==0]
cluster1 = data[data['Cluster']==1]
```

```python
plt.scatter(cluster0['x'], cluster0['y'])
plt.scatter(cluster1['x'], cluster1['y'])
```
