```python
import pandas as pd
```

```python
data = pd.read_csv("colours.csv")
data['Colours']
```

```python
colour_category = data['Colours']
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoding = LabelEncoder()
result = label_encoding.fit_transform(colour_category)
```

```python
result
```

```python
from sklearn.preprocessing import LabelBinarizer
one_hot_encoding = LabelBinarizer()
result = one_hot_encoding.fit_transform(colour_category)
```

```python
result
```
